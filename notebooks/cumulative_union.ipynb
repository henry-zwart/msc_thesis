```python
import polars as pl
```

```python
df = pl.DataFrame(
    dict(
        pid=[0, 0, 0, 1, 1, 1],
        wave=[1, 2, 3, 1, 2, 3],
        elts=[[0], [1], [0, 2, 3], [2], [], [0]],
    )
)
```

```python
new_cc13 = (
    df.join(df, how="left", on="pid")
    .filter(pl.col("wave_right") <= pl.col("wave"))
    .drop("wave_right", "elts")
    .rename({"elts_right": "elts"})
    .group_by("pid", "wave", maintain_order=True)
    .agg(pl.col("elts").list.explode().drop_nulls().unique().sort())
)
```

```python
new_cc13
```

```python
def union(xs, ys):
    return list(set(xs + ys))


df.with_columns(pl.col("elts").rolling_map(union, window_size=2))
```
