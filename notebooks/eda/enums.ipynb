```python
import polars as pl
```

```python
bears_enum = pl.Enum(["Polar", "Panda", "Brown"])
bears = pl.DataFrame(
    {"species": ["Polar", "Panda", "Brown", "Brown", "Polar"]},
    schema={"species": bears_enum},
)
bears = pl.DataFrame({"species": [0, 1, 2, 2, 0], "name": ["a", "b", "c", "d", "a"]})
print(bears)
```

```python
import pandera.polars as pa
```

```python
schema = pa.DataFrameSchema(
    {"species": pa.Column(bears_enum, coerce=True, nullable=True)}
)
```

```python
vbears = schema.validate(bears)
```

```python
vbears.filter(pl.col("species").to_physical() == 0)
```

```python
vbears.with_columns(pl.when(pl.col("name") == "a").then(None).otherwise(pl.col("name")))
```
