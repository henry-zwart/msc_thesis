```python
from pathlib import Path

import polars as pl

import matplotlib.pyplot as plt

import seaborn as sns

from climate_attitudes.extract.dataset import ClimateAttitudesDataset
from climate_attitudes.data.plotting import plot_participation_sequence_counts

ASSETS_DIR = Path("/Users/henry/data/msc_thesis/climate-attitudes")
```

```python
data = ClimateAttitudesDataset(ASSETS_DIR)
```

### Number of participants per wave

```python
fig, axe = plt.subplots(figsize=(3.5, 3), constrained_layout=True)

sns.histplot(
    data.participant.unpivot(
        index=("participant_id", "wave_joined"),
        variable_name="Wave",
        value_name="responded",
    )
    .filter(pl.col("responded"))
    .with_columns(
        pl.col("Wave").replace_strict(
            {"wave_1": 1, "wave_2": 2, "wave_3": 3, "wave_4": 4, "wave_5": 5},
            return_dtype=pl.Int64,
        )
    ),
    x="Wave",
    discrete=True,
    shrink=0.9,
)

axe.set_ylabel("Participants")
axe.spines["top"].set_visible(False)
axe.spines["right"].set_visible(False)
axe.set_title("Participants per wave")
```

### Number of new participants per wave

```python
fig, axe = plt.subplots(figsize=(3.5, 3), constrained_layout=True)
sns.histplot(data.participant, x="wave_joined", discrete=True, shrink=0.9)
axe.set_ylabel("Participants")
axe.set_xlabel("Wave")
axe.spines["top"].set_visible(False)
axe.spines["right"].set_visible(False)
axe.set_title("New participants per wave")
```

### Number of participants per participation sequence

A participation sequence comprises a boolean value for each wave, where `True` indicates that a participant has a response for the given wave, and `False` indicates otherwise.

```python
fig, axe = plot_participation_sequence_counts(data.participant)
axe.set_title("Participation sequence counts");
```
