```python
all_cols = set(
    cc_happening_4
    + cc_anthro_4
    + cc_worry_4
    + personal_harm_4
    + future_gen_4
    + fossil_fuel_4
    + renewable_energy_4
    + govt_prio_4
)
```

```python
ques_participant_types_per_wave = (
    question.filter(pl.col("item_name").is_in(all_cols))
    .group_by("item_name", "wave", maintain_order=True)
    .agg(pl.col("participant_type").sort())
    .with_columns(
        pl.when(~pl.col("participant_type").list.contains("repeating"))
        .then(pl.lit(0))
        .otherwise(
            pl.when(~pl.col("participant_type").list.contains("new"))
            .then(pl.lit(2))
            .otherwise(pl.lit(1))
        )
        .alias("participant_types")
    )
    .drop("participant_type")
    .with_columns(
        pl.col("wave").replace_strict(
            {i: f"wave_{i}" for i in range(1, 6)}, return_dtype=str
        )
    )
    .pivot("wave", values="participant_types")
    .with_columns(pl.col(r"^wave_\d$"))
)
```
