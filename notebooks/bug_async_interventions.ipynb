```python
import numpy as np
import numpy.typing as npt

from ising import Ising

RANDOM_SEED = 202606051903
```

```python
def state_configuration(
    _model: Ising, y: npt.NDArray[np.int64]
) -> npt.NDArray[np.int64]:
    return y


for pw_effect in (0.5, 1.0, 1.5, 2.0):
    model = Ising(
        field=np.array([-0.3, 0.7, 0.4]),
        coupling=np.array(
            [
                [1.0, pw_effect, pw_effect],
                [pw_effect, 1.0, 0.0],
                [0.0, pw_effect, 1.0],
            ]
        ),
        infer_structure=True,
        rng=RANDOM_SEED,
    )

    i_model = model.intervene(
        spins=np.array([1]), field_offset=np.array([10.0]), seed=RANDOM_SEED
    )

    model.reset(RANDOM_SEED)
    i_model.reset(RANDOM_SEED)

    measurements = model.measure(
        state_configuration,
        y0=np.array([-1, 1, -1]),
        t=5,
        repeats=100,
        method="sequential",
        warmup_steps=0,
        take_every=True,
    ).y[0]
    i_measurements = i_model.measure(
        state_configuration,
        y0=np.array([-1, 1, -1]),
        t=5,
        repeats=100,
        method="sequential",
        warmup_steps=0,
        take_every=True,
    ).y[0]

    print((i_measurements - measurements).mean(axis=0)[-1])
```
