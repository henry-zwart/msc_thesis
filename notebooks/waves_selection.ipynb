```python
import numpy as np
import polars as pl
import polars.selectors as cs

import matplotlib.pyplot as plt
import seaborn as sns

import networkx as nx
import iplotx as ipx

from climate_attitudes.settings import Config
from climate_attitudes.dataset import Dataset
from climate_attitudes.utils import pcorr

config = Config(_env_file="../.env")

data = Dataset.load(config)
```

```python
pids = (
    data.participant.filter("wave_2", "wave_3", "wave_4")
    .select("participant_id")
    .collect()
    .to_series()
    .implode()
)
question_cols = pl.exclude("participant_id", "wave")

df_23 = (
    data.response.filter(pl.col("participant_id").is_in(pids))
    .filter(pl.col("wave").is_in([2, 3]))
    .select(
        "participant_id",
        "wave",
        "cc1",
        "cc2",
        pl.col(r"^cc4_(world|wealthUS|poorUS|comm)$"),
        # "cc10",   # Excluded because of nulls
        "cc12",
        "cvcc_worryothers",
        "ew5",
        "ew6",
        "cc6",
        # "cc11",   # Excluded because of nulls
        # "cvcc2b",
        # "pol_interest",
        "pol_party",
        "pol_ideology",
        "ccCompensation",
        # "Variant_ccCompensation",
        "ccSolving",
        # "Variant_ccSolving",
        # "pol_vote_support",      # Excluded because of nulls
        # "Group_pol_vote_support",      # Excluded because of nulls
        "cc_ica",
        "cc_pol_tax",
        "cc_pol_car",
        "cvcc4_personal",
        "cvcc4_should",
        # "cvcc6",  # Excluded because of nulls
        "pol4",
        "pol7",
        "pol9",
        "pol11",
    )
    # .with_columns(     # Excluded because of nulls
    #     pl.col("cc2")
    #     .is_in(["Human activities", "Both"])
    #     .cast(int)
    #     .alias("cc2_anthropogenic"),
    #     pl.col("cc2")
    #     .is_in(["Natural causes", "Both"])
    #     .cast(int)
    #     .alias("cc2_natural_causes"),
    # )
    .drop(pl.col("cc2"))
    .filter(
        pl.all_horizontal(pl.all().is_not_null()),
        pl.all_horizontal(pl.col(r"^cc4_.*$") != 99),
    )
    .with_columns(pl.len().over("participant_id").alias("n_waves"))
    .filter(n_waves=2)
    .drop("n_waves")
    .with_columns(pl.col("cc1").replace({99: 1, 1: 2}))
    .with_columns(cs.enum().cast(int))
    # Re-code pol\d columns such that liberal view is higher
    .with_columns(
        -pl.col("pol4", "pol9", "pol11"),
    )
    .with_columns(
        (question_cols - question_cols.min())
        / (question_cols.max() - question_cols.min())
    )
    # .select("wave", "^Variant_.*$")
    # .with_columns(pl.col("^Variant_.*$") * 100)
    # .select(pl.all().is_null().any())
).collect()
```

```python
from climate_attitudes.utils import (
    calculate_stationary_distribution,
    calculate_proportion_response,
    calculate_transition_probabilities,
    relative_entropy,
)

plot_data = {
    "Column": [],
    "Wave": [],
    "Relative entropy": [],
    "Max eigenvalue": [],
}
for question_col in df_23.select(question_cols).columns:
    for wave in (2, 3):
        R = calculate_transition_probabilities(
            df_23, column=question_col, start=2, end=3
        )
        mu = calculate_stationary_distribution(R)
        props = calculate_proportion_response(df_23, column=question_col, wave=wave)
        eigs, _ = np.linalg.eig(R)
        plot_data["Column"].append(question_col)
        plot_data["Wave"].append(wave)
        plot_data["Relative entropy"].append(relative_entropy(props, mu))
        plot_data["Max eigenvalue"].append(np.abs(eigs[~np.isclose(eigs, 1)]).max())


plot_df = pl.DataFrame(plot_data)
```

```python
fig, axe = plt.subplots(figsize=(16, 4), constrained_layout=True)
sns.barplot(plot_df, x="Column", y="Relative entropy", hue="Wave", gap=0.2, axe=axe)

axe.tick_params(axis="x", rotation=45)
for ticklabel in axe.get_xticklabels():
    ticklabel.set_horizontalalignment("right")
```

Plot maximum non-one eigenvalue for each question.

```python
fig, axe = plt.subplots(figsize=(20, 20), constrained_layout=True)

adj = corr.values.copy() - np.eye(corr.values.shape[0])
# adj[np.abs(adj) < 0.15] = 0

dt = [("weight", float), ("corr", float)]

graph_data = np.zeros(corr.values.shape, dtype=dt)
graph_data["weight"] = 2 * np.abs(adj)
graph_data["weight"][graph_data["weight"] < 0.2] = 0
graph_data["corr"] = adj

G = nx.from_numpy_array(graph_data)

edge_linewidth = {(u, v): z["weight"] for u, v, z in G.edges(data=True)}

layout = nx.spring_layout(G, k=2)

edge_colour = [
    "tab:blue" if z["corr"] > 0 else "tab:red" for u, v, z in G.edges(data=True)
]

with ipx.style.context(
    [
        "hollow",
        {
            "edge": {
                "colour": edge_colour,
            }
        },
    ]
):
    network_artist = ipx.network(
        G,
        layout=layout,
        # tension=1,
        # edge_labels=edge_labels,
        node_labels=corr.index,
        edge_linewidth=edge_linewidth,
        edge_curved=True,
        aspect="equal",
        edge_label_bbox=dict(
            edgecolor="black",
            facecolor="white",
            boxstyle="round,pad=0.2",
        ),
        edge_label_rotate=False,
        axe=axe,
    )[0]
plt.show()
```
