# Bug: No difference between intervention and no-intervention for t=2

This is **not** a bug. This occurs when we intervene on the local field of the intervention spin, and measure the outcome at a distinct target spin. The dynamics are as follows:

$t=1$: The measured state in the intial configuration. This is prior to/at the time of intervention, so no spins' behaviour is affected yet.

$t=2$: The first sampled state. The distribution for each spin depends only on the initial configuration (from $t=1$), in combination with their local thresholds, and interaction effects. Note that:

- The intervention and no-intervention scenarios have the same initial configuration, and
- The interaction effects are unchanged in this intervention.

So any difference in the behaviour of a given spin between the intervention and no-intervention models is entirely due to a difference in local thresholds. Yet the only spin with a change in local threshold is the intervention spin. So we expect _no difference_ in behaviour at the target spin when this is distinct from the intervention spin. 

$t=3$: The behaviour of the intervention spin, realised at $t=2$, now propagates to other spins via interaction effects. Therefore we only now observe a change in the behaviour of other spins.

## Interventions on interaction effects

If we instead intervene on the interaction effects, we then observe change in target spins at $t=2$. 

However, in this case, for the **asymmetric model** we now no longer observe change in the spin at the 'sending' end of the intervened interaction until $t=3$. The reason for this is that this spin has seen no change in threshold, nor incoming interactions, and the configuration at $t=1$ is identical between intervention and no-intervention scenarios. It is only once the effects of intervening have propagated to the other spins that these effects may return to this spin.

In the **symmetric model**, on the other hand, there is no one 'sending' end of an intervention. Each intervention implicitly updates the effect in both edge directions. Therefore we expect all spins to change behaviour beginning at $t=2$.

```python
from functools import partial

import matplotlib.pyplot as plt
import numpy as np
import numpy.typing as npt

from ising import Ising

RANDOM_SEED = 202606051030
```

```python
def s_i(_model: Ising, y: npt.NDArray[np.int64], i: int) -> np.int64:
    return y[i]
```

```python
model = Ising(
    field=np.array([-0.3, 0.7, 0.4]),
    coupling=np.array(
        [
            [1.0, 0.5, 0.5],
            [0.5, 1.0, 0.0],
            [0.0, 0.5, 1.0],
        ]
    ),
    infer_structure=True,
    rng=RANDOM_SEED,
)

i_model = model.intervene(
    interactions=np.array([[1, 0]]), coupling_offset=np.array([3.0]), seed=RANDOM_SEED
)
```

```python
fig, axes = plt.subplots(figsize=(10, 4), ncols=2, constrained_layout=True)
model.draw(ax=axes[0])
i_model.draw(ax=axes[1])
axes[0].set_title("Base model")
axes[1].set_title("Intervention")
```

```python
f = partial(s_i, i=0)
model.reset(RANDOM_SEED + 1)
i_model.reset(RANDOM_SEED + 1)
measurement = model.measure(
    f, y0=np.array([-1, 1, -1]), t=10, repeats=2, warmup_steps=0
)
i_measurement = i_model.measure(
    f, y0=np.array([-1, 1, -1]), t=10, repeats=2, warmup_steps=0
)
```
