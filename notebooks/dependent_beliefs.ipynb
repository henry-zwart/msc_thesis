# Dependent beliefs

`cvcc10_cc` asks how optimistic participants are that climate change will be solved by technological solutions. We see low relation between this variable and others. However, consider a participant who does not believe climate change is happening. How would they respond to this question? Do we need to treat the cases separately?

```python
from pathlib import Path

import polars as pl
import polars.selectors as cs
import matplotlib.pyplot as plt
import seaborn as sns

from climate_attitudes.settings import Config
from climate_attitudes.dataset import Dataset
from climate_attitudes import configure_mpl

FONT_PATH = Path("../fonts")
configure_mpl(FONT_PATH)

plt.rc("figure", dpi=150)
```

```python
config = Config(_env_file="../.env")
dataset = Dataset.load(config, name="ds1")
dataset_std = dataset.standardise(cs.exclude("participant_id", "wave"))
resp = dataset.response.collect()
resp_std = dataset_std.response.collect()
```

```python
sns.displot(
    resp.filter(pl.col("cc1").is_in([0, 2])),
    x="cvcc10_cc",
    hue="cc1",
    multiple="dodge",
    discrete=True,
    shrink=0.7,
    common_norm=False,
    stat="probability",
)
```

```python
sns.heatmap(
    resp.group_by("cc4_world", "cc5_world")
    .agg(pl.len())
    .with_columns(pl.col("len") / pl.col("len").sum())
    .sort(by=("cc4_world", "cc5_world"))
    .pivot(on="cc4_world", index="cc5_world")
    .fill_null(0)
    .sort(by="cc5_world", descending=True)
    .drop("cc5_world"),
    vmin=0,
    cmap="RdBu",
)
```

```python
(
    resp.group_by("cc4_world", "cc5_world")
    .agg(pl.len())
    .with_columns(pl.col("len") / pl.col("len").sum())
    .sort(by=("cc4_world", "cc5_world"))
    .pivot(on="cc4_world", index="cc5_world")
    .fill_null(0)
    .sort(by="cc5_world", descending=True)
    .drop("cc5_world")
)
```
