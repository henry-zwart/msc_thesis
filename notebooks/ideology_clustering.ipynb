```python
import networkx as nx
import numpy as np
```

```python
J_cons = np.load("../reports/thesis/results/data/model/ideology_fit_conservative.npz")[
    "params"
][7:].reshape((7, 7))
J_lib = np.load("../reports/thesis/results/data/model/ideology_fit_liberal.npz")[
    "params"
][7:].reshape((7, 7))

J_cons[abs(J_cons) < 1e-2] = 0
J_lib[abs(J_lib) < 1e-2] = 0

J_cons[np.diag_indices_from(J_cons)] = 0
J_lib[np.diag_indices_from(J_lib)] = 0
```

```python
G_cons = nx.from_numpy_array(J_cons, create_using=nx.DiGraph)
G_lib = nx.from_numpy_array(J_lib, create_using=nx.DiGraph)
```

```python
labels = np.array(
    [
        "CC Real",
        "CC Human",
        "CC Worry",
        "CC Others Worry",
        "Weather Worry",
        "CC Impact",
        "CC Action",
    ]
)
```

```python
for comm in nx.community.greedy_modularity_communities(G_cons):
    print(labels[list(comm)])
```

```python
for comm in nx.community.greedy_modularity_communities(G_cons):
    print(labels[list(comm)])
```

```python
for comm in nx.community.greedy_modularity_communities(G_lib):
    print(labels[list(comm)])
```
